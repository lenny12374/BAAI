```python
# ============================================
# Evi Tan - Business Analytics & AI Application
# File  : bca_regression.py
# Task  : Regression Analysis for BCA (BBCA.JK)
# Date  : 2025-12-07
# ============================================

import pandas as pd
import numpy as np
import statsmodels.api as sm
from scipy import stats
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.stats.diagnostic import het_breuschpagan
from statsmodels.stats.stattools import durbin_watson
import io

pd.set_option("display.float_format", lambda x: f"{x:,.4f}")

CSV_PATH = "/content/sample_data/bca_regression_data_1.csv"   # sesuaikan jika perlu

# -------------------------------------------------------------------
# 1. LOAD & PREPARE DATA
# -------------------------------------------------------------------
def load_data(path=CSV_PATH, min_year=2015):
    """
    Load CSV BCA. File menggunakan ; sebagai separator.
    Filter minimal mulai tahun min_year.
    """
    df = pd.read_csv(path, sep=";")
    df["Year"] = pd.to_numeric(df["Year"])
    df = df[df["Year"] >= min_year].copy()
    return df

def prepare_regression_data(df):
    """
    Siapkan dataset untuk regresi:
    Dependent   : Stock_Price
    Independent : Revenue, EPS
    """
    cols_needed = ["Stock_Price", "Revenue", "EPS"]
    missing_cols = [c for c in cols_needed if c not in df.columns]
    if missing_cols:
        raise ValueError(f"Kolom berikut tidak ada di CSV kamu: {missing_cols}")

    df_reg = df[cols_needed].dropna()
    y = df_reg["Stock_Price"]
    X = df_reg[["Revenue", "EPS"]]
    X = sm.add_constant(X)
    return df_reg, X, y

# -------------------------------------------------------------------
# 2. REGRESSION & METRICS
# -------------------------------------------------------------------
def run_regression(X, y):
    model = sm.OLS(y, X).fit()
    return model

def regression_table(model):
    params = model.params
    bse = model.bse
    tvalues = model.tvalues
    pvalues = model.pvalues
    conf_int = model.conf_int(alpha=0.05)

    rows = []
    for name in params.index:
        beta = params[name]
        se = bse[name]
        tval = tvalues[name]
        pval = pvalues[name]
        ci_low, ci_high = conf_int.loc[name]
        if pval < 0.01:
            sig = "**"
        elif pval < 0.05:
            sig = "*"
        elif pval < 0.10:
            sig = "†"
        else:
            sig = "NS"
        rows.append({
            "Variable": name,
            "Coefficient": beta,
            "Std Error": se,
            "t-value": tval,
            "p-value": pval,
            "95% CI Low": ci_low,
            "95% CI High": ci_high,
            "Significance": sig
        })
    return pd.DataFrame(rows)

def compute_rmse(model, y):
    resid = model.resid
    mse = np.mean(resid**2)
    rmse = np.sqrt(mse)
    return rmse

# -------------------------------------------------------------------
# 3. ASSUMPTION TESTS
# -------------------------------------------------------------------
def run_assumption_tests(model, X, y):
    resid = model.resid
    results = {}

    # Shapiro–Wilk (normality)
    shapiro_stat, shapiro_p = stats.shapiro(resid)
    results["Shapiro_p"] = shapiro_p

    # Breusch–Pagan (homoskedastisitas)
    bp_stat, bp_p, _, _ = het_breuschpagan(resid, model.model.exog)
    results["BP_p"] = bp_p

    # Durbin–Watson (independensi)
    dw = durbin_watson(resid)
    results["DW"] = dw

    # VIF (tanpa konstanta)
    X_vif = X.drop(columns=["const"])
    vif_rows = []
    for i in range(X_vif.shape[1]):
        vif_val = variance_inflation_factor(X_vif.values, i)
        vif_rows.append({"Variable": X_vif.columns[i], "VIF": vif_val})
    vif_df = pd.DataFrame(vif_rows)

    # Cook's Distance
    influence = model.get_influence()
    cooks_d = influence.cooks_distance[0]
    cook_df = pd.DataFrame({
        "Index": y.index,
        "CookD": cooks_d
    })

    return results, vif_df, cook_df

# -------------------------------------------------------------------
# 4. MAIN
# -------------------------------------------------------------------
if __name__ == "__main__":
    # 1) Load data
    df = load_data()
    print("=== FIRST 5 ROWS ===")
    print(df.head())

    print("\n=== DESCRIPTIVE STATISTICS ===")
    print(df.describe())

    print("\n=== CORRELATION MATRIX (utama) ===")
    print(df[["Stock_Price", "Revenue", "EPS"]].corr())

    # 2) Prepare regression dataset
    df_reg, X, y = prepare_regression_data(df)
    print("\n=== REGRESSION DATASET ===")
    print(df_reg)

    # 3) Run regression
    model = run_regression(X, y)
    print("\n=== OLS REGRESSION SUMMARY ===")
    print(model.summary())

    # 4) Regression table (rapi untuk laporan)
    print("\n=== REGRESSION RESULTS TABLE ===")
    reg_tab = regression_table(model)
    print(reg_tab)

    # 5) Model fit metrics
    rmse = compute_rmse(model, y)
    print("\n=== MODEL FIT METRICS ===")
    print(f"R-squared       : {model.rsquared:,.4f}")
    print(f"Adj. R-squared  : {model.rsquared_adj:,.4f}")
    print(f"F-statistic     : {model.fvalue:,.4f}")
    print(f"F p-value       : {model.f_pvalue:,.6f}")
    print(f"RMSE            : {rmse:,.4f}")
    print(f"Residual Std Err: {np.sqrt(model.scale):,.4f}")

    # 6) Assumption tests
    results, vif_df, cook_df = run_assumption_tests(model, X, y)

    print("\n=== ASSUMPTION TESTS ===")
    print(f"Shapiro-Wilk p-value (normality)        : {results['Shapiro_p']:.4f}")
    print(f"Breusch-Pagan p-value (homoskedasticity): {results['BP_p']:.4f}")
    print(f"Durbin-Watson statistic (independence)  : {results['DW']:.4f}")

    print("\n=== VIF TABLE ===")
    print(vif_df)

    print("\n=== COOK'S DISTANCE ===")
    print(cook_df)
    print(f"\nMax Cook's Distance: {cook_df['CookD'].max():.4f}")

    print("\n=== DONE: ANALYSIS COMPLETE ===")
```

```python
# ============================================
# Evi Tan - Business Analytics & AI Application
# File  : bca_plots.py
# Task  : Visualization for BCA Regression Project
# Date  : 2025-12-07
# ============================================

import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt

CSV_PATH = "/content/sample_data/bca_regression_data_1.csv"

# -------------------------------------------------------------------
# 1. LOAD DATA
# -------------------------------------------------------------------
def load_data(path=CSV_PATH, min_year=2015):
    df = pd.read_csv(path, sep=";")
    df["Year"] = pd.to_numeric(df["Year"])
    df = df[df["Year"] >= min_year].copy()
    return df

# -------------------------------------------------------------------
# 2. LINE CHARTS
# -------------------------------------------------------------------
def plot_returns(df):
    plt.figure(figsize=(8, 5))
    plt.plot(df["Year"], df["Stock_Return"], marker="o", label="Stock Return (%)")
    plt.plot(df["Year"], df["Market_Return"], marker="o", label="Market Return (%)")
    plt.title("Stock vs Market Return (2015–2024)")
    plt.xlabel("Year")
    plt.ylabel("Return (%)")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.savefig("plot_returns_stock_vs_market.png", dpi=300)
    plt.show()

def plot_stock_price(df):
    plt.figure(figsize=(8, 5))
    plt.plot(df["Year"], df["Stock_Price"], marker="o")
    plt.title("Stock Price (2015–2024)")
    plt.xlabel("Year")
    plt.ylabel("Stock Price")
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig("plot_stock_price.png", dpi=300)
    plt.show()

def plot_revenue_net_income(df):
    if not {"Revenue", "Net_Income"}.issubset(df.columns):
        print("Kolom 'Revenue' / 'Net_Income' tidak ada, plot dilewati.")
        return
    plt.figure(figsize=(8, 5))
    plt.plot(df["Year"], df["Revenue"], marker="o", label="Revenue")
    plt.plot(df["Year"], df["Net_Income"], marker="o", label="Net Income")
    plt.title("Revenue & Net Income (2015–2024)")
    plt.xlabel("Year")
    plt.ylabel("IDR (sesuai skala)")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.savefig("plot_revenue_vs_net_income.png", dpi=300)
    plt.show()

def plot_eps(df):
    plt.figure(figsize=(8, 5))
    plt.plot(df["Year"], df["EPS"], marker="o")
    plt.title("Earnings Per Share (EPS) (2015–2024)")
    plt.xlabel("Year")
    plt.ylabel("EPS")
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig("plot_eps.png", dpi=300)
    plt.show()

# -------------------------------------------------------------------
# 3. SCATTER + REGRESSION LINE
# -------------------------------------------------------------------
def scatter_with_regression(df, xcol, ycol, filename, title=None):
    x = df[xcol]
    y = df[ycol]

    plt.figure(figsize=(6, 5))
    plt.scatter(x, y, alpha=0.8, label="Data")

    m, b = np.polyfit(x, y, 1)
    x_line = np.linspace(x.min(), x.max(), 100)
    y_line = m * x_line + b
    plt.plot(x_line, y_line, label="Linear fit")

    plt.xlabel(xcol.replace("_", " "))
    plt.ylabel(ycol.replace("_", " "))
    plt.title(title if title else f"{ycol} vs {xcol}")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.savefig(filename, dpi=300)
    plt.show()

# -------------------------------------------------------------------
# 4. RESIDUAL & Q-Q PLOTS UNTUK MODEL:
#    Stock_Price ~ Revenue + EPS
# -------------------------------------------------------------------
def residual_and_qq_plots(df):
    y = df["Stock_Price"]
    X = df[["Revenue", "EPS"]]
    X = sm.add_constant(X)

    model = sm.OLS(y, X).fit()
    fitted = model.fittedvalues
    residuals = model.resid

    # Residual vs Fitted
    plt.figure(figsize=(6, 5))
    plt.scatter(fitted, residuals, alpha=0.8)
    plt.axhline(0, linestyle="--")
    plt.xlabel("Fitted values (Ŷ)")
    plt.ylabel("Residuals (e)")
    plt.title("Residual Plot – Stock_Price ~ Revenue + EPS")
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig("plot_residuals_stock_price_model.png", dpi=300)
    plt.show()

    # Q-Q Plot
    plt.figure(figsize=(6, 5))
    sm.qqplot(residuals, line="45", fit=True)
    plt.title("Q-Q Plot of Residuals – Stock_Price Model")
    plt.tight_layout()
    plt.savefig("plot_qq_residuals_stock_price_model.png", dpi=300)
    plt.show()

# -------------------------------------------------------------------
# 5. MAIN
# -------------------------------------------------------------------
if __name__ == "__main__":
    df = load_data()
    print("Kolom yang tersedia:", list(df.columns))

    # Line charts
    plot_returns(df)
    plot_stock_price(df)
    plot_revenue_net_income(df)
    plot_eps(df)

    # Scatter plots (untuk laporan)
    scatter_with_regression(
        df,
        xcol="Revenue",
        ycol="Stock_Price",
        filename="scatter_stock_price_vs_revenue.png",
        title="Stock Price vs Revenue"
    )

    scatter_with_regression(
        df,
        xcol="EPS",
        ycol="Stock_Price",
        filename="scatter_stock_price_vs_eps.png",
        title="Stock Price vs EPS"
    )

    # Residual & Q-Q plot untuk model utama
    residual_and_qq_plots(df)
```
